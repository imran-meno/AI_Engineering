```python
#basic preprocessing
import re
reviews = [
    "This product is AMAZING!!!",
    "This product is AMAZING!!!",
    "   Very bad product   ",
    "I luv this product",
    ""
]
cleaned_reviews=[]
for review in reviews:
    #removing extra spaces
    review=review.strip()

    # making text smaller
    review= review.lower()

    #removing RE 

    review = re.sub(r"!", "", review)
    
    if not review:
        continue
    cleaned_reviews.append(review)

print(cleaned_reviews)    
```

```python
#tokenization
import re
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
text='The students are playing games, and they were running!'

text=re.sub('!','',text)

token=word_tokenize(text)
print(token)
```

```python
import string

punc=string.punctuation

other = [
    "I LOVE this product!!!",
    "The product is VERY GOOD.",
    "I don't like this product...",
    "Amazing quality, amazing price!"

 
]

cleaned_text=[]
for text in other:
    # making lower text
    text=text.lower()
    # removing puncation

    for puncation in string.punctuation:
        text=text.replace(puncation,'')
    cleaned_text.append(text)
print(cleaned_text ,end='\n')
```
