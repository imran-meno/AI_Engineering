```python

import pandas as pd
import re
import string
dataset = pd.read_csv('IMDB Dataset.csv')
df = pd.DataFrame(dataset)

cleaned_reviews = []

for review in df['review']:

    # 1. Making text lowercase
    text = review.lower()

    # 2. Removing HTML tags
    text = re.sub(r'<[^>]+>', '', text)

    # 3. Removing punctuation
    for punctuation in string.punctuation:
        text = text.replace(punctuation, '')

    # 4. Removing extra spaces from ends
    text = text.strip()

    # 5. Saving cleaned review
    cleaned_reviews.append(text)

df['review'] = cleaned_reviews


```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Use only 5 reviews first for practice
reviews = df['review'].head(5)

cv = CountVectorizer(ngram_range=(2,2))
X = cv.fit_transform(reviews)
vocubalry=cv.get_feature_names_out()
print("Matrix shape:", X.shape)
#print("Vocabulary:", cv.get_feature_names_out())

# Display only these 5 reviews as a dense array

print(X.toarray())
```
