Created the Text classification model using NLP that predict the sentiment of movie weather the given review is positive or negativa
 

```python
import pandas as pd
import re
ds=pd.read_csv('IMDB Dataset.csv')
df=pd.DataFrame(ds)

X=df['review'].iloc[:1000]
y=df['sentiment'].iloc[:1000]
cleaned_reviews=[]
for review in X:
  review=re.sub(re.compile('<.*>'),'',review)


  review=review.lower()
  cleaned_reviews.append(review)


cleaned_df=pd.DataFrame(cleaned_reviews,columns=['review'])

```

```python
#creating model

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
le=LabelEncoder()
y=le.fit_transform(y)

X_train,X_test,y_train,y_test=train_test_split(cleaned_df,y,test_size=.2,random_state=42)


#importing bow 
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer()

X_train_bow=cv.fit_transform(X_train['review']).toarray()
X_test_bow=cv.transform(X_test['review']).toarray()

#importing model

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
model=GaussianNB()
model.fit(X_train_bow,y_train)
predict=model.predict(X_test_bow)

accuracy=accuracy_score(y_test,predict)
accuracy*100
```
